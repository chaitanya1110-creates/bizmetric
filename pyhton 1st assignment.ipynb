```python
from functools import reduce

def factorial(n):
  if n == 0:
    return 1
  else:
    return reduce(lambda x, y: x * y, range(1, n + 1))

print(f"Factorial of 11: {factorial(11)}")
print(f"Factorial of 0: {factorial(0)}")
```

```python
large = lambda x, y: x if x > y else y

print(f"The larger of 17 and 5 is: {large(17, 5)}")
print(f"The larger of 30 and 81 is: {large(30, 81)}")
```

```python

area_of_circle = lambda r: 3.14 * (r**2)

radius1 = 5
area1 = area_of_circle(radius1)
print(f"The area of a circle with radius {radius1} is: {area1:.2f}")

radius2 = 10
area2 = area_of_circle(radius2)
print(f"The area of a circle with radius {radius2} is: {area2:.2f}")
```

```python
original = [9,1,2,3,44,9,1,0,12,11,1]

seen = set()
remove = lambda y: list(filter(lambda x: x not in seen and not seen.add(x), y))

unique_list = remove(original)
print(unique_list)
```
